## Data Science Task 1: Visualizing Distribution of Categorical and Continuous Variables


## Loaded the population dataset, filtered for 2022 data, and visualized the top 10 most populated countries with a bar chart.


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('API_SP.POP.TOTL_DS2_en_csv_v2_19373.csv', skiprows=4)
print(df.head())

population_2022 = df[['Country Name', '2022']].dropna()

top10 = population_2022.sort_values(by='2022', ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(top10['Country Name'], top10['2022'])
plt.xticks(rotation=45, ha='right')
plt.title('Top 10 Most Populated Countries (2022)')
plt.xlabel('Country')
plt.ylabel('Population')
plt.show()
```
